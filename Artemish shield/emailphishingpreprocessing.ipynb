```python
import pandas as pd
import csv
import sys
from google.colab import files

# --- Step 1: Prepare for Large and Complex CSV Files ---

# Increase the CSV field size limit for very large text fields.
# This prevents errors where a single cell has too much text.
print("Setting CSV field size limit...")
try:
    max_int = sys.maxsize
    while True:
        try:
            csv.field_size_limit(max_int)
            break
        except OverflowError:
            max_int = int(max_int/10)
except Exception as e:
    print(f"Warning: Could not set CSV field size limit: {e}")

# --- Step 2: Load the Dataset with Advanced Error Handling ---

print("\nLoading 'Phishing_Email.csv'...")
try:
    # Use the 'python' engine for flexibility and 'on_bad_lines' to skip problematic rows.
    # This is the key fix to prevent the script from crashing on parsing errors.
    df = pd.read_csv(
        'Phishing_Email.csv',
        engine='python',
        on_bad_lines='warn'  # This warns about and skips bad lines instead of erroring
    )
    print("Successfully loaded the dataset. Any skipped lines due to errors were reported above.")

    # --- Step 3: Clean the Data ---

    # Drop the first column (the old index) if it exists.
    if 'Unnamed: 0' in df.columns:
        df = df.drop(columns=['Unnamed: 0'])
        print("Dropped 'Unnamed: 0' column.")

    # Rename the other columns as requested.
    df = df.rename(columns={'Email Text': 'Text', 'Email Type': 'Label'})
    print("Renamed columns to 'Text' and 'Label'.")

    # --- Step 4: Save and Download the Cleaned File ---

    output_filename = 'Phishing_Email_cleaned.csv'
    df.to_csv(output_filename, index=False)
    print(f"\nCleaned data saved to '{output_filename}'.")

    # Display the first 5 rows of the final, cleaned dataframe.
    print("\n--- First 5 Rows of Cleaned Data ---")
    print(df.head())

    # Download the file in Google Colab.
    print(f"\nAttempting to download '{output_filename}'...")
    files.download(output_filename)
    print("Download initiated successfully.")

except Exception as e:
    print(f"\n--- AN UNEXPECTED ERROR OCCURRED ---")
    print(f"Error details: {e}")
    print("\nPlease double-check that the file 'Phishing_Email.csv' is uploaded to your Colab session.")
```

```python
import pandas as pd
from google.colab import files # Import the files library

# Load your two datasets
df1 = pd.read_csv('Dataset_5971_renamed_labels.csv')
df2 = pd.read_csv('Phishing_Email_cleaned.csv')

# --- Step 1: Standardize Column Names ---
# The column names for the text and label were slightly different in each file.
# This step renames the columns in the second dataframe to match the first.
print("Standardizing column names...")
df2_renamed = df2.rename(columns={'Text': 'TEXT', 'Label': 'LABEL'})

# --- Step 2: Select Only the Required Columns ---
# This ensures we only work with the 'LABEL' and 'TEXT' columns from each file.
df1_subset = df1[['LABEL', 'TEXT']]
df2_subset = df2_renamed[['LABEL', 'TEXT']]

# --- Step 3: Concatenate (Combine) the Datasets ---
# This operation stacks the rows from both datasets into a single dataframe.
print("Combining the datasets...")
combined_df = pd.concat([df1_subset, df2_subset], ignore_index=True)

# --- Step 4: Save the Result ---
# The final, combined data is saved to a new CSV file.
output_filename = 'Combined_Phishing_Dataset.csv'
combined_df.to_csv(output_filename, index=False)
print(f"Successfully combined the data and saved it to '{output_filename}'")

# --- Step 5: Download the File ---
# This is the new part that triggers the download in your browser.
print(f"\nInitiating download for {output_filename}...")
files.download(output_filename)
print("Download should start shortly!")
```

```python
import pandas as pd
import nltk
import re
import csv
import sys
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from google.colab import files

# Download necessary NLTK data
print("Downloading NLTK data...")
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
print("NLTK data download complete.")
```

```python
# --- Data Cleaning ---
print("\nCleaning text data...")

def clean_text(Text):
    # Convert to lowercase
    Text = Text.lower()
    # Remove punctuation, numbers, and special characters
    Text = re.sub(r'[^a-zA-Z\s]', '', Text)
    return Text

# Handle missing values in 'Text' column by filling with an empty string
data['Text'].fillna('', inplace=True)

data['cleaned_text'] = data['Text'].apply(clean_text)

print("Text cleaning complete.")
print("\nData with 'cleaned_text' column:")
print(data[['Text', 'cleaned_text']].head())
```

```python
# --- Tokenization ---
print("\nTokenizing cleaned text...")

data['tokens'] = data['cleaned_text'].apply(word_tokenize)

print("Tokenization complete.")
print("\nData with 'tokens' column:")
print(data[['cleaned_text', 'tokens']].head())
```

```python
# --- Stop Word Removal ---
print("\nRemoving stop words...")

stop_words = set(stopwords.words('english'))

def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

data['text_no_stopwords'] = data['cleaned_text'].apply(remove_stopwords)

print("Stop word removal complete.")
print("\nData with 'text_no_stopwords' column:")
print(data[['cleaned_text', 'text_no_stopwords']].head())
```

```python
data.head()
```

```python
data = data[['Label', 'cleaned_text']]
print("DataFrame after dropping columns:")
print(data.head())
```

```python
# --- Stemming ---
print("\nStemming text...")

stemmer = PorterStemmer()

def stem_tokens(tokens):
    return [stemmer.stem(word) for word in tokens]

data['stemmed_text'] = data['cleaned_text'].apply(stem_tokens)

print("Stemming complete.")
print("\nData with 'stemmed_text' column:")
print(data[['cleaned_text', 'stemmed_text']].head())
```

```python
data.head()
```

```python
data = data[['Label', 'cleaned_text']]
print("DataFrame after dropping columns:")
print(data.head())
```

```python
# --- Lemmatization ---
print("\nLemmatizing text...")

lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

data['lemmatized_text'] = data['cleaned_text'].apply(lemmatize_tokens)

print("Lemmatization complete.")
print("\nData with 'lemmatized_text' column:")
print(data[['cleaned_text', 'lemmatized_text']].head())
```

```python
data = data[['Label', 'cleaned_text']]
print("DataFrame after dropping columns:")
print(data.head())
```

```python
# --- Save and Download ---
print("\nSaving the fully processed dataset...")

# Creating a final processed text string for easy use
data['final_processed_text'] = data['cleaned_text'].apply(lambda tokens: ' '.join(tokens))

output_filename = 'perfectdataset_processed.csv'
data.to_csv(output_filename, index=False)

print(f"Processed data saved to '{output_filename}'.")

# Trigger the download
files.download(output_filename)
```
